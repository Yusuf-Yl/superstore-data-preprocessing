```python
import pandas as pd
import numpy as np
```

```python
#Superstore veri setini Pandas ile hafızaya yüklüyoruz
df = pd.read_csv('train.csv')
```

```python
#Verinin ilk 5 satırına bakarak verinin sağşığı yerinde mi diye bakıyoruz
df.head()
```

```python
#1. Sütun tipleri ve doluluk oranları
print("--- İNFO RAPORU ---")
print(df.info())

#2. Hangi sütunda kaç tane eksik (NaN) değer var?
print("\n--- EKSİK VERİ SAYILARI ---")
print(df.isna().sum())

#3. Tabloda birebir aynı olan tekrarlı satır var mı ?
print("\n--- TEKRARLI SATIR SAYISI ---")
print(df.duplicated().sum())
```

```python
#df.info(): Tablomuzda toplam kaç satır var, sütunların türü sayısal mı metin mi (örn: Sales sayı mı yoksa yanlışlıkla metin olarak mı kaydedilmiş) ve genel bellek kullanımını görmek için ilk baktığımız yerdir.
#df.isna().sum(): Her bir sütunda kaç tane boş (eksik) hücre olduğunu tek tek saydırıp listeler. Böylece hangi sütunu dolduracağımızı veya sileceğimizi anlarız.
#df.duplicated().sum(): Sistemde yanlışlıkla iki kez kaydedilmiş (birebir kopya) satır var mı diye arar. Eğer varsa temizlememiz gerekir ki analizimiz bozulmasın.   
```

```python
# Birebir aynı olan tekrarlı satırları tablodan siliyoruz
df=df.drop_duplicates()

#Temizlik sonrası kalan satir sayısını kontrol edelim
print("Temizlik sonrası toplam satır sayısı:" , df.shape[0])
```

```python
#Tablodaki tüm sutun isimlerini listele
print(df.columns)
```

```python
#1. Order Date sütununu gerçek tarih (datetime) formatımna çeviriyoruz
df['Order Date'] = pd.to_datetime(df['Order Date'], format='%d/%m/%Y')

#2. Tarihin içindenYıl, Ay ve Gün bilgilerini çekip yeni sütunlar olarak ekliyoruz
df['Order_Year'] = df['Order Date'].dt.year
df['Order_Month'] = df['Order Date'].dt.month
df['Order_Day'] = df['Order Date'].dt.day

#Yeni eklenen sütunlara bir bakalım 
df[['Order Date' , 'Order_Year' , 'Order_Month' , 'Order_Day']].head()
```

```python
#Postal Code sütunundaki eksik (NaN) değerleri medyan (ortanca) değeri ile doğruluyoruz
df['Postal Code'] = df['Postal Code'].fillna(df['Postal Code'].median())

#Eksik kalmış mı kontrol edelim 
print("Kalan eksik Postal Code sayısı:", df['Postal Code'].isna().sum())
```

```python
# Category ve Segment sütunlarını One-Hat Encodşng yöntemiyle 0 ve 1'lerden oluşan ayrı sütunlara çeviriyoruz
df = pd.get_dummies(df, columns=['Category', 'Segment'], drop_first=True)

# Dönüşüm sonrası tablomuzun ilk haline ve yeni sütunlara bakalım
print(df.columns)
df.head(2)
```

```python
from sklearn.preprocessing import StandardScaler

#StandardScaler modelini çağırıyoruz
scaler = StandardScaler()

# Sales sütununu ölçeklendirip yeni bir sütun olarak ekliyoruz 
df['Sales_Scaled'] = scaler.fit_transform(df[['Sales']])

# Ölçeklendirilmiş halini kontrol edelim 
df[['Sales', 'Sales_Scaled']].head()
```

```python
# Temizleniş ve dönüştürülmüş tüm DateFrame'i parquet dosyası olarak kaydediyoruz
df.to_parquet('superstore_temiz.parquet')

print("Veri başarıyla 'superstore_temiz.parquet' olarak kaydedildi!")
```
